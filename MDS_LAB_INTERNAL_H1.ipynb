```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv("sample_data/california_housing_train.csv")
x=df['median_income']
y=df['median_house_value']
plt.plot(x[:50],y[:50],color='blue',marker='o')
plt.title("Line PLot")
plt.xlabel("Income")
plt.ylabel("House Price")
plt.grid(True)
plt.show()
plt.bar(x[:50],y[:50],color='red')
plt.title("Bar Plot")
plt.xlabel("Income")
plt.ylabel("House Price")
plt.show()
plt.hist(y,color='green',edgecolor='black')
plt.title("Histogram")
plt.xlabel("House Price")
plt.ylabel("Frequency")
plt.show()
y.plot(kind='density',color='red')
plt.title("Density Plot")
plt.xlabel("Income")
plt.ylabel("House Price")
plt.show()
plt.scatter(x[:50],y[:50],color='blue',alpha=0.5)
plt.title("Scatter PLot")
plt.xlabel("Income")
plt.ylabel("House Price")
plt.show()
```
